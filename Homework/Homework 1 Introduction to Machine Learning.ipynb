### 6. Median value of horsepower

```python
initial_median = df['horsepower'].median()
horsepower_mode = df['horsepower'].mode()[0]

df['horsepower_filled'] = df['horsepower'].fillna(horsepower_mode)
new_median = df['horsepower_filled'].median()

print(f"Initial Median: {initial_median}")
print(f"New Median: {new_median}")

# comparison check
if new_median > initial_median:
    print("Answer: Yes, it increased")
elif new_median < initial_median:
    print("Answer: Yes, it decreased")
else:
    print("Answer: No")
```

### 7. Sum of weights 

```python
asia = df[df["origin"] == "Asia"]

X = asia[["vehicle_weight", "model_year"]].iloc[:7].values

XTX = X.T @ X

XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv @ X.T @ y

print(w)
print(f"{w.sum():.3f}")
```
